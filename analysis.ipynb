```python
#arrange or group by the Category and look into the profit
df.groupby("Category")["Profit"].sum()
```

```python
#arrange or group by the Category and look into the Sales
df.groupby("Category")["Sales"].sum()
```

```python
#finding total profit
profit_total= df.groupby("Category")["Profit"].sum()
```

```python
#finding Total Sales
sales_total=df.groupby("Category")["Sales"].sum()
```

```python
#finding profit per unit
profit_per_unit=profit_total/sales_total
profit_per_unit
```

```python
# going inside the furniture details and finding data
furniture = df[df["Category"] == "Furniture"]
```

```python
#grouping by sub-category and looking into the profits and finally calculating the profit sum
furniture.groupby("Sub-Category")["Profit"].sum() 
```

```python
#Finding the discount on every furniture sub-category
furniture.groupby("Sub-Category")["Discount"].mean()
```

```python
# mean of discount
df.groupby("Discount")["Profit"].mean()
```

```python
#Converting the type of date from an object to a real date
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
df["Month"] = df["Order Date"].dt.month
```

```python
df.info()
```

```python
df.groupby("Month")["Sales"].sum()
```

```python
# Created a bar chart showing Sales by Month
df.groupby("Month")["Sales"].sum().plot(kind="bar", title="Sales by Month", xlabel="Month", ylabel="Total Sales")
```

```python
# Showing a bar chart by profit by region
df.groupby("Region")["Profit"].sum().plot(kind="bar",title="Profit by Region", xlabel="Region", ylabel="Total Profit")
```

```python
df.groupby("Region")["Discount"].mean()
```

Key Findings

The discounts above 20 % lose money on average.
This is why tables and bookcases lose money and also the central region is the weakest region.

Central has the highest discount (24%) and the lowest profit. West has the lowest discount (11%) and the highest profit.

Sales peak from September to December, and February is the weakest month.
